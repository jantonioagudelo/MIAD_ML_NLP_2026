## Datos para la predicción de género en películas

![image info](https://raw.githubusercontent.com/albahnsen/MIAD_ML_and_NLP/main/images/moviegenre.png)

En este proyecto se usará un conjunto de datos de géneros de películas. Cada observación contiene el título de una película, su año de lanzamiento, la sinopsis o plot de la película (resumen de la trama) y los géneros a los que pertenece (una película puede pertenercer a más de un género). Por ejemplo:
- Título: 'How to Be a Serial Killer'
- Plot: 'A serial killer decides to teach the secrets of his satisfying career to a video store clerk.'
- Generos: 'Comedy', 'Crime', 'Horror'

La idea es que usen estos datos para predecir la probabilidad de que una película pertenezca, dada la sinopsis, a cada uno de los géneros.

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
# Importación librerías
import pandas as pd
import os
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.multiclass import OneVsRestClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import r2_score, roc_auc_score
from sklearn.model_selection import train_test_split
```

```python
# Carga de datos de archivo .csv
dataTraining = pd.read_csv('https://github.com/albahnsen/MIAD_ML_and_NLP/raw/main/datasets/dataTraining.zip', encoding='UTF-8', index_col=0)
dataTesting = pd.read_csv('https://github.com/albahnsen/MIAD_ML_and_NLP/raw/main/datasets/dataTesting.zip', encoding='UTF-8', index_col=0)
```

```python
# Visualización datos de entrenamiento
dataTraining
```

```python
# Visualización datos de test
dataTesting.head()
```

```python
# Separación de variables predictoras (X) y variable de interés (y) en set de entrenamiento y validación usandola función train_test_split
X_train, X_val, y_train, y_val = train_test_split(dataTraining[['plot']], dataTraining['genres'], test_size = 0.33, random_state = 42)
```

```python
# Definición de variables predictoras (X)
vect = CountVectorizer(max_features = 1000)

X_train_dtm = vect.fit_transform(X_train['plot'])
X_train_dtm.shape
```

```python
# Definición de variable de interés (y)
y_train = y_train.map(lambda x: eval(x))
label_bin = MultiLabelBinarizer()
y_train_genres = label_bin.fit_transform(y_train)
```

```python
y_train_genres.shape
```

```python
y_val_genres = label_bin.transform(y_val)
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV

param_dist = {'estimator__n_estimators': [50, 100, 200, 300, 400, 500], 'estimator__max_depth': [10, 15, 20, 25, 30], 'estimator__min_samples_split': [5, 10, 15, 20, 25]}

search = RandomizedSearchCV(OneVsRestClassifier(RandomForestClassifier(random_state=42)), param_dist, scoring='roc_auc_ovr', cv = 5)
```

```python
search.fit(X_train_dtm, y_train_genres)
```

```python
search.best_params_
```
